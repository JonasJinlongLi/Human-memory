```python
import numpy as np
import scipy.stats as stats
```

```python
scuccesses_basline = [29,	29,	28,	25,	23,	22,	20,	20,	20,	23,	24,	25]
successes_speedy = [24,	24,	18,	14,	15,	18,	16,	15,	18,	17,	24,	24]
N = 31
alpha = 0.05
number_total_tests = 12
alpha_bonf = alpha/number_total_tests
```

```python
alpha_bonf
```

```python
x1 = scuccesses_basline[0]
x2 = successes_speedy[0]
n1=N
n2=N
```

```python
p1=x1/n1
p2=x2/n2
```

```python
p_hat = (x1+x2)/(n1+n2)
z_obs = (p1-p2)/np.sqrt(p_hat*(1-p_hat)*(1/n1+1/n2))
```

```python
p_val = 2*(1-stats.norm.cdf(abs(z_obs)))
```

```python
p_val
```

```python
scuccesses_basline_np = np.array([29,	29,	28,	25,	23,	22,	20,	20,	20,	23,	24,	25])
successes_speedy_np = np.array([24,	24,	18,	14,	15,	18,	16,	15,	18,	17,	24,	24])
```

```python
n1 = N
n2 = N
```

```python
p1 = scuccesses_basline_np/n1
p2 = successes_speedy_np/n2

p_hat = (scuccesses_basline_np+successes_speedy_np)/(n1+n2)
z_obs = (p1-p2)/np.sqrt(p_hat*(1-p_hat)*(1/n1+1/n2))
p_val = 2*(1-stats.norm.cdf(abs(z_obs)))


for i in range(12):
    print(f'at position{i+1} z_obs: {z_obs[i]} p_val: {p_val[i]}')
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Position': range(1, 13),
    'Baseline Successes': scuccesses_basline_np,
    'Speedy Sucesses': successes_speedy_np,
    'Difference(Baseline and Speedy)': scuccesses_basline_np-successes_speedy_np,
    # 'z-statistic': z_obs,
    # 'p-value': p_val,
    'Speedy Significant Difference': p_val<alpha_bonf
})
```

```python
df
```

```python
successes_run_np = np.array([24,	22,	17,	18,	17,	13,	9,	11,	8,	14,	9,	13])
```

```python
p_run = successes_run_np/n2

p_hat = (scuccesses_basline_np+ successes_run_np)/(n1+n2)
z_obs_2 = (p1-p_run)/np.sqrt(p_hat*(1-p_hat)*(1/n1+1/n2))
p_val_2 = 2*(1-stats.norm.cdf(abs(z_obs_2)))


for i in range(12):
    print(f'at position{i+1} z_obs: {z_obs_2[i]} p_val: {p_val_2[i]}')
```

```python
df['Run Successes'] = successes_run_np
df['Difference(Baseline and Run)'] = scuccesses_basline_np-successes_run_np
# df['z--statistic'] = z_obs_2
# df['p--value'] = p_val_2
df['Run significant difference'] = p_val_2<alpha_bonf
```

```python
succeses_pause_np = np.array([25,	23,	22,	15,	19,	17,	13,	14,	9,	15,	20,	24])
```

```python
p_pause = succeses_pause_np/n2

p_hat = (scuccesses_basline_np+ succeses_pause_np)/(n1+n2)
z_obs_3 = (p1-p_pause)/np.sqrt(p_hat*(1-p_hat)*(1/n1+1/n2))
p_val_3 = 2*(1-stats.norm.cdf(abs(z_obs_3)))


for i in range(12):
    print(f'at position{i+1} z_obs: {z_obs_3[i]} p_val: {p_val_3[i]}')
```

```python
df['Pause Successes'] = succeses_pause_np
df['Difference(Baseline and Pause)'] = scuccesses_basline_np-succeses_pause_np
# df['z---statistic'] = z_obs_3
# df['p---value'] = p_val_3
df['Pause significant difference'] = p_val_3<alpha_bonf
```
