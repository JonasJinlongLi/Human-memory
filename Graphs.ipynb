```python
import random as rd
import matplotlib.pyplot as plt
```

```python

x = list(range(1, 13))

# Y-værdier (accuracy)
y = [0.968, 0.968, 0.903, 0.806, 0.742, 0.742, 0.645, 0.677, 0.677, 0.774, 0.806, 0.839]

plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o', linestyle='-', color='b', linewidth=2, markersize=6)
plt.title('Baseline for free recall', fontsize=14)
plt.xlabel('Letter number', fontsize=12)
plt.ylabel('Accuracy', fontsize=12)
plt.xticks(x)
plt.grid(True, linestyle='--', alpha=0.7)
plt.ylim(0, 1.05)
plt.tight_layout()
plt.show()
```

```python
#baseline and speed


x = list(range(1, 13))
y1 = [0.968, 0.968, 0.903, 0.806, 0.742, 0.742, 0.645, 0.677, 0.677, 0.774, 0.806, 0.839]
y2 = [0.774, 0.774, 0.581, 0.452, 0.484, 0.581, 0.516, 0.484, 0.581, 0.548, 0.774, 0.774]

plt.figure(figsize=(8, 5))
plt.plot(x, y1, marker='o', linestyle='-', color='b', linewidth=2, label='Baseline1')
plt.plot(x, y2, marker='s', linestyle='--', color='r', linewidth=2, label='Speed')
plt.title('Effect of Word Presentation Speed', fontsize=14)
plt.xlabel('Letter number', fontsize=12)
plt.ylabel('Accuracy', fontsize=12)
plt.xticks(x)
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
